```python
import kagglehub, pandas as pd
path = kagglehub.dataset_download('kartik2112/fraud-detection')
train = pd.read_csv(f'{path}/fraudTrain.csv')
test = pd.read_csv(f'{path}/fraudTest.csv')
df = pd.concat([train, test], ignore_index=True)
rows_before = len(df)  #will be used in the summary
```

## 1. Initial Inspection — Dhai


```python
print(df.shape)
df.info()
df.head()
```

**Notes:**
- 1,852,394 rows and 23 columns
- Target column: `is_fraud` (0 = legitimate, 1 = fraud)
- `trans_date_trans_time` and `dob` are stored as text, not dates
- `cc_num` and `zip` are stored as numbers, but they are IDs
- `Unnamed: 0` is an index column from the original files
- Merchant names start with `fraud_`, even in legitimate transactions
- `first`, `last`, and `street` contain personal information

# 2. Column Types - Turki

```python
df.dtypes
```

```python
# changing the trans_date_trans_time to datetime64[ns] data type
df["trans_date_trans_time"] = pd.to_datetime(df["trans_date_trans_time"])

# checking if the change was made
df["trans_date_trans_time"].dtype
```

```python
# changing the dob to datetime64[ns] data type
df["dob"] = pd.to_datetime(df["dob"])

# checking if the change was made
df["dob"].dtype
```

```python
# changing cc_num to string since its an identifier
df["cc_num"] = df["cc_num"].astype(str)

# checking if the change was made
df["cc_num"].dtype
```

```python
# changing zip to a string since its a location code, not a quantity
df["zip"] = df["zip"].astype(str)

# checking if the chnage was made
df["zip"].dtype
```

**Summary:**

The data types of 4 columns has changed ("trans_date_trans_time" , "dob" , "cc_num" , "zip"). The others columns already had their suitable data types.

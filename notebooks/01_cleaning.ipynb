## 3. Text Columns — Joud

```python
text_cols = df.select_dtypes(include=['object', 'string']).columns

for col in text_cols:
    count = (df[col].notna() & df[col].ne(df[col].str.strip())).sum()
    print(col, ':', count)
    df[col] = df[col].str.strip()
```

## 7. Outliers — Lana

   **Goal**: understand extreme values before deciding anything. In fraud detection, large amounts may be the fraud itself, so values are not removed just for being large. Only impossible values would be treated as errors.

```python
# Summary of statistics (Focusing on amt and city_pop)
df[['amt', 'city_pop']].describe().round(2)
```

**Note:** `amt` ranges from 1.00 to 28,948.90, but the median is only 47.45, so it is strongly right-skewed. `city_pop` ranges from 23 to 2,906,700; large cities are real, so high values are expected.

```python
# How many outliers are there?
# IQR rule: values above Q3 + 1.5 * IQR are counted as outliers
# This only counts them; nothing is removed
for col in ['amt', 'city_pop']:
    q1, q3 = df[col].quantile([0.25, 0.75])
    upper = q3 + 1.5 * (q3 - q1)
    n = (df[col] > upper).sum()
    print(f"{col}: upper bound = {upper:,.2f}, outliers = {n:,} ({n / len(df):.2%})")
```

**Note:** The IQR rule flags 95,054 `amt` values (5.13%) and 346,191 `city_pop` values (18.69%) as outliers. These are too many to be errors; they reflect the skewed distributions, so nothing is removed here.

```python
import numpy as np
import matplotlib.pyplot as plt

fig, axes = plt.subplots(1, 2, figsize=(14, 4))

# Left: all amounts; the y-axis is on a log scale so rare large amounts are visible
axes[0].hist(df['amt'], bins=100, log=True, color='mediumorchid')
axes[0].set_title('Transaction amount distribution')
axes[0].set_xlabel('amt')
axes[0].set_ylabel('Number of transactions (log scale)')

# Right: legitimate vs fraud amounts, on a log x-axis to spread out the skewed values
# Each bar shows the share of its group, so the two groups are comparable
legit = df.loc[df['is_fraud'] == 0, 'amt']
fraud = df.loc[df['is_fraud'] == 1, 'amt']
bins = np.logspace(np.log10(df['amt'].min()), np.log10(df['amt'].max()), 80)
axes[1].hist(legit, bins=bins, alpha=0.5, color='blue', weights=np.ones(len(legit)) / len(legit), label='Legitimate')
axes[1].hist(fraud, bins=bins, alpha=0.7, color='red', weights=np.ones(len(fraud)) / len(fraud), label='Fraud')
axes[1].set_xscale('log')
axes[1].set_title('Amount: legitimate vs fraud')
axes[1].set_xlabel('amt (log scale)')
axes[1].set_ylabel('Share of transactions')
axes[1].legend()

plt.tight_layout()
plt.show()
```

**Note:** Most transactions are small, with a long tail reaching about 29,000. Legitimate transactions peak around 50–100, while most fraud is concentrated around 300 and 1,000. A smaller group of fraud cases falls around 10–20, which blends in with normal spending, so amount alone can't identify all fraud.

```python
# Large amount vs fraud
# Fraud rate for amounts above 1000 (True) vs at or below 1000 (False)
# count shows how many transactions are in each group, mean is the fraud rate

df.groupby(df['amt'] > 1000)['is_fraud'].agg(['count', 'mean'])
```

**Note:** Only 5,519 transactions are above 1,000, but 22.20% of them are fraud, compared with 0.46% for the rest, about 49 times higher. Removing large amounts would remove the cases we want to detect, so they are kept.

```python
# Finding impossible values:

# Amounts and city populations can't be zero or negative
print("Amounts <= 0:", (df['amt'] <= 0).sum())
print("City population <= 0:", (df['city_pop'] <= 0).sum())

# Customer age at the time of each transaction (not saved to df)
age = (df['trans_date_trans_time'] - df['dob']).dt.days / 365.25
print("Age range:", round(age.min(), 1), "to", round(age.max(), 1))
print("Negative ages:", (age < 0).sum())
print("Ages over 100:", (age > 100).sum())
```

**Note:** No zero or negative amounts or city populations were found. Customer ages range from 13.9 to 96.2 years, so there are no impossible ages. The youngest customers are teenagers, which is possible for cards held by young account holders.

**Decision:** No rows were removed. The extreme values in `amt` and `city_pop` are real, and large amounts are closely linked to fraud. The dataset still has 1,852,394 rows and 19 columns.

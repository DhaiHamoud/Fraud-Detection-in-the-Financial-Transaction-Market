```python
import kagglehub, pandas as pd
path = kagglehub.dataset_download('kartik2112/fraud-detection')
train = pd.read_csv(f'{path}/fraudTrain.csv')
test = pd.read_csv(f'{path}/fraudTest.csv')
df = pd.concat([train, test], ignore_index=True)
rows_before = len(df)  #will be used in the summary
```

**Notes:**
- 1,852,394 rows and 23 columns
- Target column: `is_fraud` (0 = legitimate, 1 = fraud)
- `trans_date_trans_time` and `dob` are stored as text, not dates

- `first`, `last`, and `street` contain personal information

# 2. Column Types - Turki

```python
df.dtypes
```

```python
# changing the trans_date_trans_time to datetime64[ns] data type
df["trans_date_trans_time"] = pd.to_datetime(df["trans_date_trans_time"])

# checking if the change was made
df["trans_date_trans_time"].dtype
```

```python
# changing the dob to datetime64[ns] data type
df["dob"] = pd.to_datetime(df["dob"])

# checking if the change was made
df["dob"].dtype
```

```python
# changing cc_num to string since its an identifier
df["cc_num"] = df["cc_num"].astype(str)

# checking if the change was made
df["cc_num"].dtype
```

```python
# changing zip to a string since its a location code, not a quantity
df["zip"] = df["zip"].astype(str)

# checking if the chnage was made
df["zip"].dtype
```

```python
df.info()
```

**Summary:**

The data types of 4 columns has changed ("trans_date_trans_time" , "dob" , "cc_num" , "zip"). The others columns already had their suitable data types.

## 3. Text Columns — Joud

```python
text_cols = df.select_dtypes(include=['object', 'string']).columns

for col in text_cols:
    count = (df[col].notna() & df[col].ne(df[col].str.strip())).sum()
    print(col, ':', count)
    df[col] = df[col].str.strip()
```

Note: No leading or trailing spaces were found in the text columns. I applied strip() to keep the values consistent.

```python
print(df['gender'].value_counts(dropna=False))
print(df['category'].value_counts(dropna=False))
```

Note: Gender contains only F and M. The 14 category names are consistent, so no changes were needed.

```python
df['merchant'].str.startswith('fraud_').mean()
```

```python
df['merchant'] = df['merchant'].str.removeprefix('fraud_')
df['merchant'].head()
```

Note: All merchant names started with 'fraud_', including those in legitimate transactions. It is part of the name, not a fraud label. I removed it to make merchant names clearer.

## 4. Unnecessary Columns — Rayan

```python
# Step 1: Inspect the unnecessary index column

print("Column exists:", 'Unnamed: 0' in df.columns)

if 'Unnamed: 0' in df.columns:
    print("\nFirst five values:")
    print(df['Unnamed: 0'].head())

    print("\nNumber of unique values:")
    print(df['Unnamed: 0'].nunique())

    print("\nTotal rows:", len(df))
```

**Note:** `Unnamed: 0` is an index column with repeated values after merging the datasets. It is not useful for fraud detection, so we will remove it.

```python
# Remove the unnecessary index column
df.drop(columns=['Unnamed: 0'], inplace=True)

# Check the result
print("Column exists:", 'Unnamed: 0' in df.columns)
print("Dataset shape:", df.shape)
```

**Note:** The `Unnamed: 0` column was successfully removed. The number of columns decreased from 23 to 22, while the number of rows remained unchanged.

```python

# Step 2: Check unique values in each column
df.nunique()
```

**Note:** No constant columns were found. All columns have more than one unique value, so no columns were removed in this step.

```python
# Step 3: Inspect personal information columns
df[['first', 'last', 'street']].head()
```

**Note:** The `first`, `last`, and `street` columns contain personal information that is not needed for our analysis. Other location columns are available, so we will remove these three columns.

```python
# Remove unnecessary personal information columns
df.drop(columns=['first', 'last', 'street'], inplace=True)

# Check the result
print("Dataset shape:", df.shape)
print(df.columns)
```

**Note:** Successfully removed `first`, `last`, and `street`. The dataset now has 19 columns, with all rows preserved.
